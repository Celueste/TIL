#### 4장 1강 : 조건문

- 판별식에 따라서 참(True)면 특정 참에 실행되는 코드를 정의하고, 거짓(False)이면 거짓 조건에 실행되는 코드를 정의할 수 있는 문장
- 판별식: 비교 연산자, 논리 연산자, 값

1. 단일 조건문: if

```
if 판별식: 
    판별식이 참일때 실행되는 코드
```

````python
num=10
if num > 5:
    print("5보다 크다")
````

2. if-else 조건문

````python
num = 5
if num ==10:
    print("10 입니다.")
else: # if num == 10 : 거짓일때 실행되는 코드를 정리
    print("10이 아닙니다")
````

3. 다중 조건 분기: if-elif-else

````python
age = int(input())
if age >=20:
    print("성인")
elif age >= 17:
    print("고등학생")
elif age >=14:
    print("중학생")
elif age >= 8:
    print("초등학생")
else: #모든 조건문이 False 이면 실행되는 구문
    print("유치원생")
    
print(age, type(age))
````

4. 복잡한 조건 처리: 중첩 조건문과 논리 연산자 결합

````python
# 입력 받은 숫자가 10 이상 100 미만 인지 확인
num = int(input())
if num >= 10:
    if num < 100:
        print(f"{num}는 10 이상 100 미만입니다.")
````

````python
# 논리 연산자 and 결합
num = int(input())

if num >=10 and num < 100:
    print(f"{num}는 10 이상 100 미만입니다.")
````
