````python
def print_n_times(n,*values):
    # n번 반복합니다.
    for i in range(n):
        # values는 리스트처럼 활용합니다.
        for value in values:
            print(value)
        # 단순한 줄바꿈
        print()

# 함수를 호출합니다.
print_n_times(3,"하잉","즐겁고","슬픈","파이썬","공부시간")
````

````python
# 기본 매개변수: 값을 넣어주지 않았을 때 자동으로 사용할 기본값이 정해져 있는 매개변수
def hello(name="철수"):
    print(name)

hello()  # 철수

# 가변 매개변수: 몇 개의 값을 받을지 정해져 있지 않은 매개변수야.
def print_values(*values):
    print(values)

print_values("사과", "바나나", "딸기")  #('사과', '바나나', '딸기')

# 기본 매개변수가 가변 매개변수보다 앞에 올때
def print_n_times(n=2,*values):
    # n번 반복합니다.
    for i in range(n):
        # values는 리스트처럼 활용합니다.
        for value in values:
            print(value)
        # 단순한 줄바꿈
        print()

# 함수를 호출합니다.
print_n_times("하잉","즐겁고","슬픈","파이썬","공부시간")

# >> 기본 매개변수는 가변 매개변수 앞에 써도 의미가 없음
````

````python
# 가변 매개변수가 기본 매개변수보다 앞에 올때
def print_n_times(*values, n=2):
    # n번 반복합니다.
    for i in range(n):
        # values는 리스트처럼 활용합니다.
        for value in values:
            print(value)
        # 단순한 줄바꿈
        print()

# 함수를 호출합니다.
print_n_times("하잉","즐겁고","슬픈","파이썬","공부시간", 3)

# >> 가변 매개변수가 우선되어 하잉부터 3까지 2번 출력됨. 3번 출력되게 하려면 키워드 매개변수 사용해야함
````

````python
# 키워드 매개변수
def print_n_times(*values, n=2):
    # n번 반복합니다.
    for i in range(n):
        # values는 리스트처럼 활용합니다.
        for value in values:
            print(value)
        # 단순한 줄바꿈
        print()

# 함수를 호출합니다.
print_n_times("하잉","즐겁고","슬픈","파이썬","공부시간", n=3)
````

````python
def test(a, b=10, c=10):
    print(a+b+c)

# 기본 형태
test(10, 20 , 30) 

# 키워드 매개변수로 모든 매개변수를 지정한 형태
test(a=10, b=100, c=200)

# 키워드 매개변수로 모든 매개변수를 마구잡이로 지정한 형태
test(c=10, a=100, b=200)

# 키워드 매개변수로 일부 매개변수만 지정한 형태
test(10, c=200)
````

````python
# input() 함수의 리턴값을 변수에 저장
value = input(">")

# 출력
print(value)

# 함수 내부에서는 return 키워드를 사용할 수 있음. 이 키워드는 함수를 실행했던 위치로 돌아가라는 뜻으로, 함수가 끝나는 위치를 의미함.

def return_test():
    print("A 위치입니다.")
    return  # 리턴합니다.
    print("B 위치입니다.")

return_test()

함수 내부에서 출력을 두 번 했지만, 중간에 리턴 키워드가 있어서 함수가 종료되어 결과적으로 A 위치입니다만 출력하고 프로그램이 종료됨.
````

````python
# 함수를 선언합니다.
def sum_all(star, end):
    #  변수를 선언합니다.
    output = 0
    # 반복문을 돌려 숫자를 더합니다.
    for i in range(star, end+1):
        output += i
    # 리턴합니다.
    return output

# 함수를 호출합니다.
print("0 to 100: ", sum_all(0, 100))
````

````python
# 기본 매개변순와 키워드 매개변수를 활용해 범위의 정수를 더하는 함수
# 함수를 선언합니다.
def sum_all(start=0, end=100, step=1):
    # 변수를 선언합니다.
    output = 0
    # 반복문을 돌려 숫자를 더합니다.
    for i in range(start, end +1, step):
        output +=i
    # 리턴합니다.
    return output

# 함수를 호출합니다.
print(sum_all(0,100,10))
print(sum_all(end = 100))
print(sum_all(end = 100, step =2))
````
